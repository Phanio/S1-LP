. Le sens de l'objectif inversé maximiser et non minmiser
. Le sens du contrainte de la ressource feuille : <br/>
$2a_2 + a_3 ≤ 200$ et non $ ≥ 200$ <br/>
. La définition du domaine des variables : <br/>
$a_2$, $a_3$ ≥ 0

maximiser z = $4a_2 + 2a_3$ <br/>
les contraintes: <br/>
$2a_2 + a_3 ≤ 200$ <br/>

$a_2$, $a_3$ ≥ 0

Non, ce sont deux problèmes séparés : 

1. La modèle de coût, c(q) = k·q^(−θ) est une fonction de puissance: c'une courbe donc une fonction non linéaire. C'est un problème d'optimisation on cherche à trouver les paramètre k et θ à partir des données, par régression log-log et descente de gradient. Si on l'intégre dans l'objectif, le programme ne serait plus linéaire
2. Le mobèle objectif z = 120u + 200c + 175s est une fonction liénaire: un problème de décision qui cherche combien produire de chaque modèle de vélos afin de maximiser la marche sous les contraintes de ressources.

Étape 2, le programme linéaire : on utilise ces marges comme des chiffres fixes dans l’objectif: 120€ marge d'un vélo Urban, 
200€ marge d'un vélo Cargo,
175€ marge d'un vélo Sport,

Q3: <br/>
La solution optimale est : <br/>
u = 75 <br/>
c = 30 <br/>
max z = $120u + 200c$ = 120*75 + 200*30 = 15000<br/>

Ressources saturées :
- s1 = 0 -> La main-d'œuvre est entièrement consommée (4 × 75 + 6 × 30 = 480 h sur 480 h).   
- s2 = 0 -> L'aluminium est entièrement consommé (6 × 75 + 15 × 30 = 900 kg sur 900 kg).   
- Le temps machine n'est pas saturé (s3 = 30) : on utilise 2 × 75 + 2 × 30 = 210 h sur les 240 h disponibles (il reste 30 h de capacité). <br/>  

Interprétation économique : <br>
Les prix marginaux lus sur la dernière ligne indiquent : 
- y1 = 25 €/h : une heure supplémentaire de main-d'œuvre rapporterait 25 €.
- y2 = 10/3 ≈ 3,33 €/kg : un kilogramme d'aluminium supplémentaire rapporterait environ 3,33 €.
- y3 = 0 €/h : une heure machine supplémentaire n'apporte aucun gain marginal car la machine est sous-utilisée, une heure de machine ferait 31h de non utilisation (en ces heures il n'y aura pas d'ouvrier).

Q4:  
Différence méthodologique :
- A 3 produits (u, c, s), le problème passe en dimension 3 et le nombre de variables ainsi que de combinaisons de bases augmente. 
- Les calculs manuels de pivot deviennent longs, fastidieux et sujets aux erreurs d'arithmétique.   

Intérêt d'un solveur (PuLP / SciPy) :   
- Rapidité et fiabilité : résolution quasi instantanée sans risque d'erreur humaine. 
- Capacité à gérer des dizaines de contraintes et de produits si la gamme s'élargit encore.
- Analyses de sensibilité automatiques : extraction directe des prix duaux, des coûts réduits et des plages de validité des ressources pour orienter les décisions de l'entreprise.
