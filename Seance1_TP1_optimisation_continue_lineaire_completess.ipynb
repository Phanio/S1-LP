## Exercice 1 — Descente de gradient à partir de zéro

Rappel de la règle de mise à jour :
$$ \mathbf{w}_{k+1} = \mathbf{w}_k - \alpha \nabla J(\mathbf{w}_k). $$

**1.1** Complétez `gradient_descent` ci-dessous : à chaque itération, calculez le gradient de `cost_fun` au point courant (utilisez `g`, déjà construit avec `autograd`), arrêtez-vous si sa norme passe sous `eps`, sinon mettez à jour `w` et ajoutez le nouveau point à `history`.

```python
def gradient_descent(cost_fun, w_init, alpha, epochs, eps=1e-12):
    '''Minimise cost_fun par descente de gradient.

    ENTREES
    cost_fun : callable, fonction de coût J(w)
    w_init   : array-like, point initial
    alpha    : float, pas (learning rate)
    epochs   : int, nombre maximal d'itérations
    eps      : float, arrêt si la norme du gradient passe sous eps

    SORTIES
    w       : point final (np.array)
    history : liste des itérés successifs (y compris w_init)
    '''
    w = np.array(w_init, dtype=float)
    history = [w.copy()]
    g = grad(cost_fun)   # g(w) renvoie le gradient de cost_fun en w (autograd)

    for k in range(epochs):
        gk = np.array(g(w), dtype=float)        # 1) gradient au point courant
        if np.linalg.norm(gk) < eps:             # 2) critère d'arrêt
            break
        w = w - alpha * gk                       # 3) pas de descente
        history.append(w.copy())                 # 4) mémorisation de l'itéré

    return w, history
```

**✅ Vérification 1.1** — testez `gradient_descent` sur un problème 1D trivial dont vous connaissez la solution exacte : $J(w) = (w-3)^2$, minimum en $w=3$.

```python
def _J1d(w):
    return (w[0] - 3.0) ** 2

_w_check, _hist_check = gradient_descent(_J1d, w_init=[0.0], alpha=0.1, epochs=500, eps=1e-12)
assert abs(_w_check[0] - 3.0) < 1e-3, f"attendu w*≈3.0, obtenu {_w_check[0]}"
assert len(_hist_check) > 1, "history doit contenir plusieurs points"
print(f"OK : convergé en {len(_hist_check)-1} itérations vers w* = {_w_check[0]:.5f}")
```

**1.2** Appliquez la descente de gradient au bol quadratique
$$ J(w_1, w_2) = 3(w_1 - 1)^2 + (w_2 + 2)^2 $$
en partant de $w_0 = (4, 3)$, pour $\alpha = 0.05$ **puis** $\alpha = 0.3$. La fonction `show_history` ci-dessous (déjà écrite) trace les lignes de niveau de $J$ et la trajectoire des itérés — complétez juste la boucle qui appelle `gradient_descent` pour chaque `alpha` et affiche le résultat.

```python
def J_quad(w):
    return 3 * (w[0] - 1) ** 2 + (w[1] + 2) ** 2


def show_history(history, cost_fun, xlim, ylim, title=""):
    a_values, b_values = np.meshgrid(
        np.linspace(*xlim), np.linspace(*ylim)
    )
    Z = np.array(
        [cost_fun([a, b]) for a, b in zip(a_values.flat, b_values.flat)]
    ).reshape(a_values.shape)
    plt.figure()
    plt.contour(a_values, b_values, Z, levels=20, cmap="viridis")
    h = np.array(history)
    plt.plot(h[:, 0], h[:, 1], "o-", color="crimson", markersize=3)
    plt.plot(h[0, 0], h[0, 1], "ks", label="départ")
    plt.plot(h[-1, 0], h[-1, 1], "k*", markersize=12, label="arrivée")
    plt.legend()
    plt.title(title)
    plt.xlabel("$w_1$")
    plt.ylabel("$w_2$")
    plt.show()


for alpha in [0.05, 0.3]:
    w_star, hist = gradient_descent(J_quad, w_init=[4.0, 3.0], alpha=alpha, epochs=60)
    print(f"alpha = {alpha}: w* = {np.round(w_star, 4)} après {len(hist)-1} itérations, J = {J_quad(w_star):.2e}")
    show_history(hist, J_quad, xlim=(-1, 5), ylim=(-4, 4), title=f"Descente de gradient, alpha = {alpha}")
```

**Vos observations :**

La Hessienne de $J$ est diagonale : $H = \mathrm{diag}(6, 2)$. Chaque coordonnée évolue donc indépendamment, avec un facteur de contraction $|1 - \alpha \lambda_i|$ par itération :

| | direction $w_1$ ($\lambda = 6$) | direction $w_2$ ($\lambda = 2$) |
|---|---|---|
| $\alpha = 0.05$ | $1 - 0.3 = 0.7$ | $1 - 0.1 = 0.9$ |
| $\alpha = 0.3$ | $1 - 1.8 = -0.8$ | $1 - 0.6 = 0.4$ |

- Avec $\alpha = 0.05$, les deux facteurs sont **positifs** : on descend toujours du même côté, la trajectoire est lisse (mais lente selon $w_2$, facteur 0.9).
- Avec $\alpha = 0.3$, le facteur selon $w_1$ est **négatif** ($-0.8$) : à chaque pas on **dépasse** le minimum selon $w_1$ et on repart de l'autre côté, d'où le **zigzag**. Ça converge encore car $|-0.8| < 1$, mais on est proche de la limite de stabilité $\alpha < 2/\lambda_{\max} = 1/3$.

C'est l'effet du **conditionnement** $\kappa = \lambda_{\max}/\lambda_{\min} = 3$ : le pas est limité par la direction la plus courbée ($w_1$), alors que la direction la plus plate ($w_2$) voudrait un grand pas. Plus $\kappa$ est grand, plus ce compromis est mauvais.


## Exercice 2 — Méthodes accélérées : momentum & Nesterov

**2.1** Implémentez le **momentum (heavy-ball)** :
$$ \mathbf{v}_{k+1} = \beta \mathbf{v}_k - \alpha \nabla J(\mathbf{w}_k), \qquad \mathbf{w}_{k+1} = \mathbf{w}_k + \mathbf{v}_{k+1} $$
et le **gradient accéléré de Nesterov** :
$$ \mathbf{v}_{k+1} = \beta \mathbf{v}_k - \alpha \nabla J(\mathbf{w}_k + \beta \mathbf{v}_k), \qquad \mathbf{w}_{k+1} = \mathbf{w}_k + \mathbf{v}_{k+1} $$
Les deux se codent dans **une seule** fonction `momentum_gd` : le paramètre `nesterov` choisit si le gradient est évalué en `w` (momentum) ou au point "anticipé" `w + beta*v` (Nesterov).

```python
def momentum_gd(cost_fun, w_init, alpha, beta, epochs, eps=1e-12, nesterov=False):
    w = np.array(w_init, dtype=float)
    v = np.zeros_like(w)
    history = [w.copy()]
    g = grad(cost_fun)
    for k in range(epochs):
        lookahead = w + beta * v if nesterov else w   # 1) point d'évaluation du gradient
        gk = np.array(g(lookahead), dtype=float)       # 2) gradient
        if np.linalg.norm(gk) < eps:                   # 3) critère d'arrêt
            break
        v = beta * v - alpha * gk                      # 4) mise à jour de la vitesse
        w = w + v                                      #    puis de la position
        history.append(w.copy())                       # 5) mémorisation
    return w, history
```

**✅ Vérification 2.1** — même test que pour l'Exercice 1 (doit converger vers $w=3$, avec et sans Nesterov).

```python
_w1, _h1 = momentum_gd(_J1d, w_init=[0.0], alpha=0.1, beta=0.5, epochs=500, eps=1e-10, nesterov=False)
_w2, _h2 = momentum_gd(_J1d, w_init=[0.0], alpha=0.1, beta=0.5, epochs=500, eps=1e-10, nesterov=True)
assert abs(_w1[0] - 3.0) < 1e-2, f"momentum : attendu w*≈3.0, obtenu {_w1[0]}"
assert abs(_w2[0] - 3.0) < 1e-2, f"nesterov : attendu w*≈3.0, obtenu {_w2[0]}"
print(f"OK : momentum -> w*={_w1[0]:.4f} en {len(_h1)-1} it. ; nesterov -> w*={_w2[0]:.4f} en {len(_h2)-1} it.")
```

**2.2** Comparez, sur le même quadratique mal conditionné $J\_quad$ ($\alpha = 0.15$), le nombre d'itérations nécessaires pour atteindre $\|\nabla J\| < 10^{-6}$ pour : descente de gradient simple, momentum ($\beta=0.8$), Nesterov ($\beta=0.8$). Complétez la boucle `for name, fn in [...]` (donnez à chaque `fn` un appel à `gradient_descent` ou `momentum_gd` avec les bons paramètres) ; le tracé du diagramme en barres est déjà écrit.

```python
results = {}
for name, fn in [
    ("Descente de gradient", lambda: gradient_descent(J_quad, [4.0, 3.0], 0.15, 500, eps=1e-6)),
    ("Momentum", lambda: momentum_gd(J_quad, [4.0, 3.0], 0.15, 0.8, 500, eps=1e-6)),
    ("Nesterov", lambda: momentum_gd(J_quad, [4.0, 3.0], 0.15, 0.8, 500, eps=1e-6, nesterov=True)),
]:
    w_star, hist = fn()
    results[name] = len(hist) - 1
    print(f"{name:22s}: {len(hist)-1:4d} itérations -> w* = {np.round(w_star, 4)}")

plt.figure()
plt.bar(results.keys(), results.values(), color=["#5B7FB5", "#E3946A", "#8CB369"])
plt.ylabel("itérations pour converger")
plt.title("Effet de l'accélération (alpha=0.15)")
plt.show()


# Expérience suggérée : même comparaison sur un bol beaucoup plus mal conditionné
def make_J(a):
    return lambda w: a * (w[0] - 1) ** 2 + (w[1] + 2) ** 2

print("\nInfluence du conditionnement (alpha choisi = 1/L, beta = 0.9) :")
for a in [3, 30, 300]:
    J_a = make_J(a)
    alpha_a = 1.0 / (2 * a)            # 1/L avec L = 2a, la plus grande valeur propre de la Hessienne
    n_gd = len(gradient_descent(J_a, [4.0, 3.0], alpha_a, 20000, eps=1e-6)[1]) - 1
    n_mo = len(momentum_gd(J_a, [4.0, 3.0], alpha_a, 0.9, 20000, eps=1e-6)[1]) - 1
    n_ne = len(momentum_gd(J_a, [4.0, 3.0], alpha_a, 0.9, 20000, eps=1e-6, nesterov=True)[1]) - 1
    print(f"  kappa = {a:3d} : GD = {n_gd:5d} it. | momentum = {n_mo:4d} it. | Nesterov = {n_ne:4d} it.")
```

**Vos observations :**

Résultats obtenus avec $\alpha = 0.15$ : descente de gradient **46** itérations, momentum **146**, Nesterov **52**.

Oui, c'est surprenant : **ici, le momentum est plus lent que la descente de gradient simple.** Deux raisons :

- Le problème est **bien conditionné** ($\kappa = 3$). La descente de gradient converge déjà vite : son facteur de contraction est $\max(|1-0.15\times6|, |1-0.15\times2|) = 0.7$ par itération.
- Avec $\beta = 0.8$, la « vitesse » accumulée est trop grande : l'itéré **dépasse** le minimum et oscille autour de lui avant de se poser. Le taux de convergence du heavy-ball est de l'ordre de $\sqrt{\beta} \approx 0.89$, donc moins bon que 0.7. Pour $\kappa = 3$, le $\beta$ optimal théorique est $\left(\frac{\sqrt\kappa - 1}{\sqrt\kappa + 1}\right)^2 \approx 0.07$, très loin de 0.8.
- Nesterov corrige en partie ce défaut : il évalue le gradient au point anticipé $w + \beta v$, ce qui « freine » avant de dépasser. Il fait presque aussi bien que la descente simple.

**Quand le conditionnement se dégrade, la tendance s'inverse** (deuxième tableau ci-dessus, $\alpha = 1/L$, $\beta = 0.9$) :

| $\kappa$ | Descente de gradient | Momentum | Nesterov |
|---|---|---|---|
| 3 | 40 | 290 | 58 |
| 30 | 476 | 319 | 188 |
| 300 | 4 828 | 322 | 245 |

Le nombre d'itérations de la descente simple croît comme $\kappa$, alors que celui des méthodes accélérées croît seulement comme $\sqrt{\kappa}$. **L'accélération n'est utile que sur des problèmes mal conditionnés**, ce qui est justement le cas courant en apprentissage.


## Exercice 3 — Optimisation continue sous contraintes

Considérons
$$ \min_{x_1,x_2} \; (x_1 - 3/2)^2 + (x_2 - 1/2)^4 \quad \text{s.c.} \quad |x_1| + |x_2| \le 1 .$$

**3.1** Résolvez-le de **trois façons indépendantes**, une cellule à la fois, et vérifiez qu'elles s'accordent.

**(a) Barrière logarithmique.** L'idée : remplacer les 4 contraintes linéaires équivalentes à $|x_1|+|x_2|\le 1$ (à savoir $g_i(x)\ge 0$ pour $g_1=1-x_1-x_2$, $g_2=1+x_1+x_2$, $g_3=1-x_1+x_2$, $g_4=1+x_1-x_2$) par une pénalité $-\xi\sum_i \log g_i(x)$ ajoutée à l'objectif, et résoudre un problème **non contraint** avec `scipy.optimize.minimize` (méthode Nelder-Mead).

```python
import cvxpy as cp
from scipy.optimize import minimize

def f0(x1, x2):
    return (x1 - 1.5) ** 2 + (x2 - 0.5) ** 4

def f0_barrier(x, xi=0.05):
    x1, x2 = x
    g = [1 - x1 - x2, 1 + x1 + x2, 1 - x1 + x2, 1 + x1 - x2]
    if min(g) <= 0:              # hors du domaine faisable : pénalité énorme
        return 1e6
    return f0(x1, x2) - xi * np.sum(np.log(g))

sol_barrier = minimize(f0_barrier, x0=[0.0, 0.0], args=(0.01,), method="Nelder-Mead")
print("solution barrière :", sol_barrier.x, "f0 =", f0(*sol_barrier.x))
```

**(b) SLSQP avec contraintes explicites.** Cette fois, on passe directement les 4 contraintes $g_i(x)\ge 0$ à `scipy.optimize.minimize(..., method="SLSQP", constraints=...)`, sans les transformer en pénalité.

```python
cons = (
    {"type": "ineq", "fun": lambda x: 1 - x[0] - x[1]},   # g1 >= 0
    {"type": "ineq", "fun": lambda x: 1 + x[0] + x[1]},   # g2 >= 0
    {"type": "ineq", "fun": lambda x: 1 - x[0] + x[1]},   # g3 >= 0
    {"type": "ineq", "fun": lambda x: 1 + x[0] - x[1]},   # g4 >= 0
)
sol_slsqp = minimize(lambda x: f0(x[0], x[1]), x0=[0.0, 0.0], method="SLSQP", constraints=cons)
print("solution SLSQP    :", sol_slsqp.x, "f0 =", f0(*sol_slsqp.x))
```

**(c) `cvxpy`.** Le problème est convexe : formulez-le directement en `cvxpy` (variables, contrainte `cp.abs(x1) + cp.abs(x2) <= 1`, objectif, `prob.solve()`).

```python
x1 = cp.Variable()
x2 = cp.Variable()
constraints = [cp.abs(x1) + cp.abs(x2) <= 1]
obj = cp.Minimize(cp.square(x1 - 1.5) + cp.power(x2 - 0.5, 4))
prob = cp.Problem(obj, constraints)
prob.solve()
print("solution cvxpy    :", x1.value, x2.value, "f0 =", prob.value)
```

**✅ Vérification 3.1** — les trois méthodes doivent tomber sur (à peu près) le même point.

```python
_pts = {
    "barrière": np.array(sol_barrier.x),
    "SLSQP": np.array(sol_slsqp.x),
    "cvxpy": np.array([x1.value, x2.value]),
}
for name, p in _pts.items():
    print(f"  {name:10s}: {np.round(p, 4)}")
_ref = _pts["SLSQP"]
for name, p in _pts.items():
    assert np.linalg.norm(p - _ref) < 0.05, f"{name} s'écarte trop de SLSQP : {p} vs {_ref}"
print("OK : les trois méthodes s'accordent (à 0.05 près).")
```

**Remarque.** La solution de la barrière, environ (0.976, 0.010), est légèrement à l'intérieur du losange, alors que le vrai optimum est le sommet (1, 0). C'est normal : la pénalité $-\xi \sum \log g_i$ tend vers $+\infty$ au bord, donc elle **repousse** la solution vers l'intérieur. L'écart diminue quand $\xi \to 0$. En pratique, les méthodes de points intérieurs résolvent une suite de problèmes avec $\xi$ de plus en plus petit.

À l'optimum $x^\star = (1, 0)$, on a $\nabla f_0(x^\star) = (2(1-1.5),\ 4(0-0.5)^3) = (-1,\ -0.5)$. Les contraintes actives sont $g_1$ et $g_3$, et le système KKT
$$ \begin{pmatrix}-1 & -1\\ -1 & 1\end{pmatrix}\begin{pmatrix}\mu_1\\ \mu_3\end{pmatrix} = \begin{pmatrix}-1\\ -0.5\end{pmatrix} $$
donne, en additionnant puis en soustrayant les deux lignes : $-2\mu_1 = -1.5$, donc $\mu_1 = 0.75$, et $2\mu_3 = 0.5$, donc $\mu_3 = 0.25$. Les deux sont positifs : les conditions KKT sont vérifiées.


**3.2 — Multiplicateurs de Lagrange à la main.** Écrivez les quatre contraintes sous la forme $g_i(x) \ge 0$ :
$$ g_1 = 1-x_1-x_2,\quad g_2 = 1+x_1+x_2,\quad g_3 = 1-x_1+x_2,\quad g_4 = 1+x_1-x_2 . $$
Identifiez numériquement quelles $g_i$ sont actives (nulles) à l'optimum — vous devriez trouver **deux** contraintes actives simultanément, car l'optimum se situe exactement sur un *sommet* du losange $|x_1|+|x_2|\le 1$ (le même phénomène qu'un sommet optimal du polytope d'un PL !). Sur une feuille, écrivez la condition de stationnarité KKT avec les deux multiplicateurs correspondants,
$$ \nabla f_0(x^\star) = \mu_1 \nabla g_1(x^\star) + \mu_3 \nabla g_3(x^\star), \qquad \mu_1,\mu_3 \ge 0, $$
résolvez le système linéaire 2×2 obtenu **à la main**, puis codez le même calcul ci-dessous pour vérifier votre résultat papier.

```python
x_star = np.array(sol_slsqp.x)
g = [1 - x_star[0] - x_star[1], 1 + x_star[0] + x_star[1],
     1 - x_star[0] + x_star[1], 1 + x_star[0] - x_star[1]]
print("valeurs des contraintes g1..g4 :", np.round(g, 4), " (0 = active)")

grad_f0 = np.array([2 * (x_star[0] - 1.5), 4 * (x_star[1] - 0.5) ** 3])
M = np.array([[-1, -1],
              [-1,  1]])          # colonnes : grad(g1) = (-1,-1) et grad(g3) = (-1, 1)
mu = np.linalg.solve(M, grad_f0)
print("grad f0(x*) =", np.round(grad_f0, 4))
print("mu1, mu3 =", np.round(mu, 4), " (les deux doivent être >= 0 : complémentarité vérifiée)")
```

**✅ Vérification 3.2** — la stationnarité KKT doit être satisfaite (résidu quasi nul) et les deux multiplicateurs doivent être positifs.

```python
_grad_g1, _grad_g3 = np.array([-1, -1]), np.array([-1, 1])
_residual = grad_f0 - (mu[0] * _grad_g1 + mu[1] * _grad_g3)
assert np.linalg.norm(_residual) < 1e-4, f"stationnarité KKT non vérifiée, résidu = {_residual}"
assert mu[0] > -1e-6 and mu[1] > -1e-6, f"les deux multiplicateurs doivent être >= 0, obtenu {mu}"
print(f"OK : résidu KKT = {np.linalg.norm(_residual):.2e}, mu = {np.round(mu, 4)} (>= 0)")
```

## Exercice 4 — Programmation linéaire : mix de production

Avant l'extension de sa gamme avec le modèle **Sport** (voir le Projet, Partie 1 ci-dessous, pour la gamme réelle à trois produits), ExedBike ne fabriquait que deux modèles de vélos électriques : **Urban** (marge 120€/unité) et **Cargo** (marge 200€/unité) — c'est cette version simplifiée, déjà résolue à la main en TD1 (Partie ExedBike — Résolution), que nous reprenons ici pour la vérifier avec un solveur.
Chaque semaine, l'usine dispose de **480 heures de main-d'œuvre**, **900 kg d'aluminium pour batteries**, et **240 heures-machine**.

| Ressource            | Urban | Cargo | Disponible |
|---------------------|:----:|:----:|:---:|
| Main-d'œuvre (h/unité)      | 4    | 6    | 480 |

**4.1a** Formulez et résolvez le PL avec `scipy.optimize.linprog` (rappel : `linprog` **minimise** — pour maximiser la marge, minimisez son opposé).

```python
from scipy.optimize import linprog

c = [-120, -200]                 # on minimise -(120 u + 200 c)
A_ub = [[4, 6],                  # main-d'oeuvre
        [6, 15],                 # aluminium
        [2, 2]]                  # temps machine
b_ub = [480, 900, 240]

res = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=(0, None), method="highs")
print("scipy.optimize.linprog :")
print("  Urban =", round(res.x[0], 2), " Cargo =", round(res.x[1], 2),
      " marge hebdomadaire = ", round(-res.fun, 2), "€")
```

**4.1b** Formulez le **même** PL avec `PuLP` (langage plus proche de l'écriture mathématique : variables, objectif à *maximiser* directement, contraintes une par une).

```python
import pulp

prob = pulp.LpProblem("ExedBike_mix", pulp.LpMaximize)
u = pulp.LpVariable("Urban", lowBound=0)
cg = pulp.LpVariable("Cargo", lowBound=0)

prob += 120 * u + 200 * cg, "marge_totale"
prob += 4 * u + 6 * cg <= 480, "main_oeuvre"
prob += 6 * u + 15 * cg <= 900, "aluminium"
prob += 2 * u + 2 * cg <= 240, "temps_machine"

prob.solve(pulp.PULP_CBC_CMD(msg=False))
print("Statut PuLP :", pulp.LpStatus[prob.status])
print("Urban =", u.value(), " Cargo =", cg.value(), " marge = ", pulp.value(prob.objective), "€")

for name, con in prob.constraints.items():
    print(f"  {name:14s} marge de manoeuvre = {con.slack:.2f}  (saturée : {abs(con.slack) < 1e-6})")
```

### 1. Ajuster un modèle de coût de production (optimisation continue)

Les données ci-dessous forment un jeu de données synthétique du coût unitaire de production en fonction du volume cumulé produit pour le modèle Urban (un effet classique de **courbe d'apprentissage** : le coût baisse avec l'expérience accumulée). Le modèle est une loi de puissance,
$$ c(q) = k\, q^{-\theta}, \qquad \theta \in (0,1). $$
En passant au logarithme, ceci devient une **régression linéaire** : $\log c(q) = \log k - \theta \log q$.

```python
# Jeu de données synthétique (ne changez pas la graine : les résultats doivent être reproductibles)
rng = np.random.default_rng(7)
q = np.linspace(50, 2000, 40)
k_true, theta_true = 9500.0, 0.55
c_obs = k_true * q ** (-theta_true) * np.exp(rng.normal(0, 0.04, size=q.shape))

plt.figure()
plt.scatter(q, c_obs, s=15, label="coût unitaire observé")
plt.xlabel("volume cumulé (unités)")
plt.ylabel("coût unitaire (EUR)")
plt.title("ExedBike Urban — données de courbe d'apprentissage")
plt.legend()
plt.show()
```

Ajustez $(\log k, \theta)$ en minimisant l'erreur quadratique moyenne **avec votre propre implémentation de la descente de gradient** (celle de l'Exercice 1, réutilisable telle quelle) sur les données log-log **standardisées**, puis reportez $(k,\theta)$ dans les unités d'origine. Tracez les données et l'ajustement, ainsi que la courbe de convergence, et expliquez (dans votre synthèse d'une page) pourquoi standardiser $\log q$ compte ici (faites le lien avec le mauvais conditionnement observé à l'Exercice 1.2).

Étapes suggérées :
1. `x = log(q)`, `y = log(c_obs)` ;
2. standardisez `x` : `x_z = (x - x.mean()) / x.std()` ;
3. définissez `J_fit(params)` = EQM entre `p0 - p1*x_z` et `y` (avec `params=[p0,p1]`, en utilisant `anp` pour que `autograd` puisse dériver) ;
4. `gradient_descent(J_fit, w_init=[0.0, 0.0], alpha=..., epochs=...)` ;
5. dé-standardisez : `theta_hat = p1/x_std`, `k_hat = exp(p0 + p1*x_mean/x_std)` ;
6. tracez le nuage de points + la courbe $k\_hat \cdot q^{-\theta\_hat}$, et la courbe de convergence de la perte.

```python
x = np.log(q)
y = np.log(c_obs)

# 2) standardisation de x = log q
x_mean, x_std = x.mean(), x.std()
x_z = (x - x_mean) / x_std

# 3) perte : erreur quadratique moyenne du modèle linéaire y ≈ p0 - p1 * x_z
def J_fit(params):
    p0, p1 = params[0], params[1]
    return anp.mean((p0 - p1 * x_z - y) ** 2)

# 4) descente de gradient (notre implémentation de l'Exercice 1)
#    Sur x_z standardisé, la Hessienne vaut 2*diag(1, 1) : conditionnement parfait,
#    donc on peut prendre un grand pas (alpha < 2/L = 1).
w_fit, hist = gradient_descent(J_fit, w_init=[0.0, 0.0], alpha=0.4, epochs=1000, eps=1e-10)
p0_hat, p1_hat = w_fit

# 5) retour aux unités d'origine
#    log c = p0 - p1 (log q - mean)/std = (p0 + p1*mean/std) - (p1/std) log q
theta_hat = p1_hat / x_std
k_hat = np.exp(p0_hat + p1_hat * x_mean / x_std)

print(f"convergé en {len(hist)-1} itérations")
print(f"ajustement : k = {k_hat:.1f}, theta = {theta_hat:.3f}   (vraies valeurs : k = {k_true}, theta = {theta_true})")

# Contrôle : solution exacte des moindres carrés sur les données brutes
slope, intercept = np.polyfit(x, y, 1)
print(f"contrôle polyfit : k = {np.exp(intercept):.1f}, theta = {-slope:.3f}")

# 6) graphiques
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
q_grid = np.linspace(q.min(), q.max(), 300)
axes[0].scatter(q, c_obs, s=15, label="coût unitaire observé")
axes[0].plot(q_grid, k_hat * q_grid ** (-theta_hat), color="crimson",
             label=f"ajustement $c(q) = {k_hat:.0f}\\,q^{{-{theta_hat:.3f}}}$")
axes[0].set_xlabel("volume cumulé (unités)")
axes[0].set_ylabel("coût unitaire (EUR)")
axes[0].set_title("Courbe d'apprentissage ajustée")
axes[0].legend()

losses = [J_fit(w) for w in hist]
axes[1].semilogy(losses)
axes[1].set_xlabel("itération")
axes[1].set_ylabel("EQM (échelle log)")
axes[1].set_title("Convergence de la descente de gradient")
plt.tight_layout()
plt.show()

# Comparaison sans standardisation : la Hessienne est très mal conditionnée
def J_raw(params):
    return anp.mean((params[0] - params[1] * x - y) ** 2)

H_raw = 2 * np.array([[1, -x.mean()], [-x.mean(), np.mean(x ** 2)]])
eig = np.linalg.eigvalsh(H_raw)
alpha_raw = 1.9 / eig.max()          # le plus grand pas stable
_, hist_raw = gradient_descent(J_raw, [0.0, 0.0], alpha_raw, 5000, eps=1e-10)
print(f"\nSans standardisation : conditionnement de la Hessienne = {eig.max()/eig.min():.0f}")
print(f"  -> {len(hist_raw)-1} itérations (plafond 5000) et EQM finale = {J_raw(hist_raw[-1]):.4f}"
      f" contre {J_fit(w_fit):.4f} avec standardisation")
```

**Résultats et pourquoi standardiser.**

L'ajustement donne $\hat k \approx 9\,683$ et $\hat\theta \approx 0{,}555$, très proches des vraies valeurs ($k = 9\,500$, $\theta = 0{,}55$), et identiques à la solution exacte des moindres carrés (`np.polyfit`). L'écart avec les vraies valeurs vient du bruit des données, pas de l'algorithme.

**Pourquoi la standardisation compte.** La perte $J(p_0, p_1) = \frac1n\sum (p_0 - p_1 x_i - y_i)^2$ est un bol quadratique dont la Hessienne vaut $2\begin{pmatrix}1 & -\bar x\\ -\bar x & \overline{x^2}\end{pmatrix}$.

- Sur $x = \log q$ brut, $\bar x \approx 6{,}6$ : les deux paramètres sont fortement couplés et le **conditionnement vaut environ 2 900**. On retrouve exactement la situation de l'Exercice 1.2, en bien pire : le pas est limité par la direction la plus courbée, et l'algorithme avance très lentement dans la direction la plus plate. Résultat mesuré : 5 000 itérations sans converger.
- Sur $x_z$ standardisé (moyenne 0, écart-type 1), la Hessienne devient $2I$ : **conditionnement égal à 1**, les lignes de niveau sont des cercles, et le gradient pointe directement vers le minimum. Résultat : **16 itérations**.

Standardiser ne change pas le modèle (on revient aux unités d'origine à la fin), cela change seulement la géométrie du problème vue par l'algorithme.


### 2. PL du mix de production pour la gamme réelle

ExedBike vend en réalité **trois** modèles : Urban, Cargo et Sport, avec les données ci-dessous.
```python
produits = ["Urban", "Cargo", "Sport"]
marges = {"Urban": 120, "Cargo": 200, "Sport": 175}
ressources = {
    "main_oeuvre":   ({"Urban": 4, "Cargo": 6,  "Sport": 5}, 520),
    "aluminium":     ({"Urban": 6, "Cargo": 15, "Sport": 9}, 1100),
    "temps_machine": ({"Urban": 2, "Cargo": 2,  "Sport": 3}, 260),
}

# 1) variables de décision
mix = pulp.LpProblem("ExedBike_mix_3_produits", pulp.LpMaximize)
xv = {p: pulp.LpVariable(p, lowBound=0) for p in produits}

# 2) objectif
mix += pulp.lpSum(marges[p] * xv[p] for p in produits), "marge_totale"

# 3) contraintes de ressources
for r, (conso, dispo) in ressources.items():
    mix += pulp.lpSum(conso[p] * xv[p] for p in produits) <= dispo, r

# 4) résolution
mix.solve(pulp.PULP_CBC_CMD(msg=False))

# 5) résultats
print("Statut :", pulp.LpStatus[mix.status])
print("Mix optimal (unités / semaine) :")
for p in produits:
    print(f"  {p:6s} = {xv[p].value():7.2f}")
print(f"Marge totale = {pulp.value(mix.objective):.2f} €\n")

print(f"{'ressource':14s} {'utilisé':>9s} {'dispo':>7s} {'écart':>7s} {'saturée':>8s} {'prix dual':>10s}")
for r, con in mix.constraints.items():
    dispo = ressources[r][1]
    print(f"{r:14s} {dispo - con.slack:9.2f} {dispo:7d} {con.slack:7.2f} {str(abs(con.slack) < 1e-6):>8s} {con.pi:8.2f} €")

# Coûts réduits : combien chaque produit "perd" par rapport à la valeur des ressources qu'il consomme
print("\nCoûts réduits :")
for p in produits:
    print(f"  {p:6s} : {xv[p].dj:7.2f} €/unité")

# Contrôle avec scipy
res3 = linprog([-120, -200, -175], A_ub=[[4, 6, 5], [6, 15, 9], [2, 2, 3]],
               b_ub=[520, 1100, 260], bounds=(0, None), method="highs")
print(f"\nContrôle scipy : x = {np.round(res3.x, 2)}, marge = {-res3.fun:.2f} €, "
      f"prix duaux = {np.round(-res3.ineqlin.marginals, 2)}")

# Version entière (on produit des vélos entiers)
mix_int = pulp.LpProblem("ExedBike_mix_entier", pulp.LpMaximize)
xi = {p: pulp.LpVariable(p + "_int", lowBound=0, cat="Integer") for p in produits}
mix_int += pulp.lpSum(marges[p] * xi[p] for p in produits)
for r, (conso, dispo) in ressources.items():
    mix_int += pulp.lpSum(conso[p] * xi[p] for p in produits) <= dispo
mix_int.solve(pulp.PULP_CBC_CMD(msg=False))
print("Version entière :", {p: int(xi[p].value()) for p in produits},
      f"marge = {pulp.value(mix_int.objective):.0f} €")
```
